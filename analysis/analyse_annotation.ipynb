```python
import pandas as pd
from collections import defaultdict, Counter
```

```python
data = pd.read_csv("/home/knf792/gits/MultimodalTaxonomy/results/vlm_outputs_gemma3.csv")
data["labels"] = data["labels"].apply(lambda x: eval(x))
data["label_set"] = data["labels"].apply(lambda labels: set([l["label"] for l in labels]))
```

```python
all_labesl = [l for labels in data["label_set"] for l in labels]

counts = Counter(all_labesl)
for key, val in counts.most_common(20):
    print(f"{key}: {val / len(data) * 100:.2f}%")
```
